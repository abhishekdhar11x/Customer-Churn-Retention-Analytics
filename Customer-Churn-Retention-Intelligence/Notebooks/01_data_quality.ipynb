```python
df["TotalCharges"] = df["TotalCharges"].replace(" ", "0")
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])

df["TotalCharges"].dtype
```

```python
df["TotalCharges"].isnull().sum()

df["TotalCharges"].dtype

df[df["tenure"] == 0][["tenure", "MonthlyCharges", "TotalCharges", "Churn"]]
```

```python
for column in df.select_dtypes(include="object").columns:
    print(column)
    print(df[column].unique())
    print()
```

```python
df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0})

df["SeniorCitizen"] = df["SeniorCitizen"].map({1: "Yes", 0: "No"})

df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 6, 12, 24, 48, 72],
    labels=["0-6 Months", "7-12 Months", "13-24 Months", "25-48 Months", "49-72 Months"]
)

df["MonthlyRevenue"] = df["MonthlyCharges"]

df["RevenuePerTenureMonth"] = df["TotalCharges"] / df["tenure"].replace(0, 1)
```

```python
df.to_csv("/content/telco_customer_churn_cleaned.csv", index=False)
```
